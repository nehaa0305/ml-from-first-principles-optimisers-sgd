```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
def run_optimizer(f,grad_f,lr,theta,iterations=100):
 num_samples=100
 x_values=np.linspace(-10,10,num_samples)
 y_values=np.linspace(-10,10,num_samples)
 X,Y=np.meshgrid(x_values,y_values)
 Z=f(X,Y)
 trajectory=[theta.copy()]
 losses=[]
 grad_norms=[]
 for i in range(iterations):
  x,y=theta
  grad=grad_f(x,y)
  theta=theta-lr*grad
  trajectory.append(theta.copy())
  losses.append(f(theta[0],theta[1]))
  grad_norms.append(np.linalg.norm(grad))
 trajectory=np.array(trajectory)
 plt.figure(figsize=(6,6))
 plt.contour(X,Y,Z,levels=30)
 plt.plot(trajectory[:,0],trajectory[:,1],'r-o',markersize=3)
 plt.scatter(0,0,color='black',s=80)
 plt.title("Optimization Path")
 plt.show()
 plt.plot(losses)
 plt.title("Loss vs Iterations")
 plt.show()
 plt.plot(grad_norms)
 plt.title("Gradient Norm vs Iterations")
 plt.show()



```

```python
#let us take a simple convex function x**2+y**2

def f1(x,y):return x**2+y**2
def g1(x,y):return np.array([2*x,2*y])


```

```python
run_optimizer(f1,g1,lr=0.1,theta=np.array([-8.0,8.0]))
```
